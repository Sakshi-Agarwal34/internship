```python
import numpy as np

arr = np.array([[1, np.nan, 3],
                [4, 5, np.nan],
                [7, 8, 9]])

# Column-wise mean (NaN को ignore करके)
col_mean = np.nanmean(arr, axis=0)

# NaN की positions
inds = np.where(np.isnan(arr))

# NaN को column mean से replace करना
arr[inds] = np.take(col_mean, inds[1])

print(arr)
```
