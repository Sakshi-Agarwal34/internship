```python
import pandas as pd

 #original dataframe
df = pd.DataFrame({
    "name":["ram","radha","krishna"],
    "age":[20,21,22]
})

print("Original DataFrame:")
print(df)

# new row
new_row = pd.DataFrame({
    "name":["mohan"],
    "age":[25]
})

# position where row will be inserted
pos = 1

# insert row
df = pd.concat([
    df.iloc[:pos],
    new_row,
    df.iloc[pos:]
]).reset_index(drop=True)

print("\nAfter inserting row:")
print(df)
```
