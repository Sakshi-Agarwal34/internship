```python

import pandas as pd
one=pd.DataFrame({"nAme":["a","b","c","d"],"subject":["sub1","sub2","sub3","sub4"],"id":[1,2,3,4]})

two=pd.DataFrame({"nAme":["x","b","y","d"],"subject":["sub1","sub4","sub5","sub6"],"id":[1,2,3,5]})

re=pd.concat([one,two],ignore_index=True)
print(re)
two=pd.DataFrame({"nAme":["x","b","y","d","e"],"subject":["sub1","sub2","sub5","sub6","sub7"],"id":[1,6,3,4,5]})
result=re.merge(two,on="id")
print(result)
#"""merge() and join() both combine two datasets, but their concept is different.

#merge()
#Used to combine data based on common columns.
#It compares values in columns and matches related rows.
#Similar to how databases (SQL) combine tables.
#More powerful and flexible.
#Example Idea

#If one table has:

#Student ID and Name

#and another has:

#Student ID and Marks

#merge() matches rows using the common column Student ID.

#join()
#Used mainly to combine data based on the index (row labels).
#It attaches one dataset to another according to row positions/indexes.
#Simpler and mainly used when indexes are already related.
#Example Idea

#If two tables have the same row indexes:

#First table → Names
#Second table → Marks

#join() combines them row by row using those indexes.
```

```python
#df.join()
It is a DataFrame method.
Mainly used to join DataFrames using their indexes.
Simpler and shorter syntax.
Best when the data is already indexed properly.
Characteristics
Joins on index by default.
Usually used for combining columns side-by-side.
Faster and easier for simple index-based operations.
Concept

“Attach another table to the current table using row labels (indexes).”

#pd.merge()
It is a pandas function.
Used to combine DataFrames using one or more common columns.
Similar to SQL joins.
More flexible and powerful.
Characteristics
Can join on columns, indexes, or both.
Supports complex joins.
Better for relational datasets.
Concept

“Match rows between tables based on common values.”
```
