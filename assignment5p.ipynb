```python
numbers = (12, 45, 7, 89, 23)

total = 0
highest = numbers[0]
lowest = numbers[0]

for num in numbers:
    total += num

    if num > highest:
        highest = num

    if num < lowest:
        lowest = num

print("Total =", total)
print("Highest =", highest)
print("Lowest =", lowest)
```

```python
n = (3, 14, 7, 22, 9, 41, 18, 5)

result = ()

for num in n:
    if num > 10:
        result += (num,)

print(result)
```
